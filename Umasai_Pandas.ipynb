##### Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 

The dataset contains several data-quality issues, including missing marks, duplicate records, inconsistent department names, and marks stored as text. The cleaning process involves inspecting the dataset, standardizing department names, removing duplicate records, converting marks into numeric values, investigating missing data, and identifying marks outside the valid range of 0–100.

**1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().** 

```python
import pandas as pd
df = pd.read_csv("student_assessment_dirty.csv")
print("First five rows:")
print(df.head())
print("\nDataset information:")
df.info()
print("\nMissing values:")
print(df.isnull().sum())
```

##### Explanation:
- The Pandas library is imported to perform data-cleaning operations, and the CSV file is loaded into a DataFrame.
- `head()` displays the first five rows, `info()` provides information about columns and data types, and `isnull().sum()` counts missing values in each column.

**2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.**  

```python
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({"I.T.": "IT"})
)
print(df["Department"].value_counts())
```

##### Explanation:
Department names are converted to uppercase, and I.T. is replaced with IT. This ensures that IT, it, and I.T. are represented consistently. Finance variations are also standardized as FINANCE.

**3.Identify and remove exact duplicate student assessment records.**  

```python
original_count = len(df)
df = df.drop_duplicates().copy()
records_removed = original_count - len(df)
print("Records removed:", records_removed)
```

##### Explanation: 
`drop_duplicates()` removes exact duplicate records. The difference between the original and cleaned row counts gives the number of records removed.

**4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.**

```python
marks_text = (
    df["Marks"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s*marks\s*$", "", regex=True, case=False)
)
df["Marks"] = pd.to_numeric(
    marks_text.replace({
        "Absent": pd.NA,
        "NA": pd.NA,
        "": pd.NA
    }),
    errors="coerce"
)
```

##### Explanation:
The text " marks" is removed from entries such as "88 marks". The remaining values are converted into numeric form, while Absent, NA, blank entries, and other `unconvertible values` become missing numeric values.

**5. Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.**  

```python
df["Attendance_Status"] = "Marks entered"
df.loc[
    marks_text.str.casefold().eq("absent").fillna(False),
    "Attendance_Status"
] = "Absent"
missing_mask = (
    marks_text.isna()
    | marks_text.eq("")
    | marks_text.str.casefold().eq("na").fillna(False)
)
df.loc[missing_mask, "Attendance_Status"] = "Marks not entered"
print(df[["Student_ID", "Student_Name", "Marks", "Attendance_Status"]])
```

##### Explanation:
The Attendance_Status column distinguishes students recorded as absent from those whose marks have not been entered. This prevents different reasons for missing marks from being treated as identical.

> Risk: Deleting missing values without investigation may remove useful student records and introduce bias into the analysis.

**6.Identify marks outside the valid range of 0–100 and decide how they should be handled.**

```python
invalid_mask = df["Marks"].notna() & (
    (df["Marks"] < 0) | (df["Marks"] > 100)
)

invalid_marks = df.loc[invalid_mask]

print("Invalid marks:")
print(invalid_marks[["Student_ID", "Student_Name", "Marks"]])

df["Invalid_Marks"] = invalid_mask
```

##### Explanation:
Marks `below 0 or above 100` are identified as invalid. These records are flagged for investigation instead of being automatically deleted, because they may contain data-entry errors.

**7.Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.**

```python
print("\nCleaned DataFrame:")
print(df)
print("\nRecords removed:", records_removed)
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:", invalid_mask.sum())
```

### Overall Explanation For Section D
- **Department names:** Convert names to uppercase and standardize I.T. as IT. This also standardizes Finance and finance as FINANCE.

- **Duplicates:** drop_duplicates() removes exact duplicate records.

- **Marks conversion:** Removes the text "marks" from values such as "88 marks" and converts the column to numeric. Absent, NA, and blank entries are treated as missing numeric marks.

- **Missing values:** The Attendance_Status column distinguishes students marked absent from students whose marks were not entered.

- **Invalid marks:** Values below 0 or above 100 are flagged for verification. They should not be deleted automatically because they may represent data-entry errors.

- **Risk of deleting missing rows:** It can remove useful student records and introduce bias if missing marks are related to absence or incomplete data entry.
